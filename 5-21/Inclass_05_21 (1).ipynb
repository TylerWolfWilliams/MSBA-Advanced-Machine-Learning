### Importing padded cat and dog images

```python
# Re-import images with aspect-ratio preservation (padded instead of raw)

train_ds_padded = tf.keras.utils.image_dataset_from_directory(
    data_dir,                                   # root folder containing class subfolders
    validation_split=0.2,                       # use 20% of data for validation
    subset="training",                          # select the training portion
    seed=42,                                    # ensures consistent train/validation split
    image_size=(128, 128),                      # target size for all images
    batch_size=32,                              # number of images per batch
    pad_to_aspect_ratio=True                    # preserve image proportions and pad instead of stretching
)

val_ds_padded = tf.keras.utils.image_dataset_from_directory(
    data_dir,                                   # same dataset location
    validation_split=0.2,                       # same split ratio
    subset="validation",                        # select the validation portion
    seed=42,                                    # must match training split
    image_size=(128, 128),                      # same target size
    batch_size=32,                              # same batch size
    pad_to_aspect_ratio=True                    # apply same preprocessing to validation data
)

# Split validation into validation + test (50/50 split)
val_batches = tf.data.experimental.cardinality(val_ds_padded).numpy()

test_ds_padded = val_ds_padded.take(val_batches // 2)
val_ds_padded = val_ds_padded.skip(val_batches // 2)

print("Train batches:", tf.data.experimental.cardinality(train_ds_padded).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds_padded).numpy())
print("Test batches:", tf.data.experimental.cardinality(test_ds_padded).numpy())
```

```python
random.seed(42)

cat_paths = list(Path(data_dir).glob("cats_set/*"))
dog_paths = list(Path(data_dir).glob("dogs_set/*"))

image_paths = random.sample(cat_paths, 3) + random.sample(dog_paths, 3)
random.shuffle(image_paths)

plt.figure(figsize=(6, 12))

for i, path in enumerate(image_paths):
    img = tf.io.read_file(str(path))
    img = tf.image.decode_jpeg(img, channels=3)

    raw_img = tf.image.resize(img, (128, 128))
    padded_img = tf.image.resize_with_pad(img, 128, 128)

    label = path.parent.name

    plt.subplot(6, 2, 2*i + 1)
    plt.imshow(raw_img.numpy().astype("uint8"))
    plt.title(f"Raw: {label}")
    plt.axis("off")

    plt.subplot(6, 2, 2*i + 2)
    plt.imshow(padded_img.numpy().astype("uint8"))
    plt.title(f"Padded: {label}")
    plt.axis("off")

plt.suptitle("Same Images: Raw Resize vs Padded Resize")
plt.tight_layout()
plt.show()
```

### Normalize all data and shuffle train data

```python
# Normalize pixel values from 0–255 to 0–1 for all datasets
normalization_layer = tf.keras.layers.Rescaling(1./255)

train_ds_padded = train_ds_padded.map(
    lambda x, y: (normalization_layer(x), y)     # normalize images; keep labels unchanged
)

val_ds_padded = val_ds_padded.map(
    lambda x, y: (normalization_layer(x), y)     # apply same normalization to validation data
)

test_ds_padded = test_ds_padded.map(
    lambda x, y: (normalization_layer(x), y)     # apply same normalization to test data
)

# Shuffle only the training data to improve learning
# Did above but best practice to do after all preprocessing steps
train_ds_padded = train_ds_padded.shuffle(
    1000,                                        # buffer size (>= dataset size for good mixing)
    seed=42                                      # ensures reproducible shuffling
)
```

```python
# check normalization worked correctly (should be 0–1 range)
for images, labels in train_ds_padded.take(1):
    print(images.numpy().min(), images.numpy().max())
```

### Building the MLP model

```python
# set random seed for reproducibility
tf.random.set_seed(42)

# Define the MLP model
mlp_catdog = models.Sequential([
    layers.Input(shape=(128, 128, 3)),       # input is a 128x128 color image
    layers.Flatten(),                        # convert image to a 1D vector (loses spatial structure)
    layers.Dense(128, activation="relu"),    # hidden layer learns patterns from pixel values
    layers.Dense(64, activation="relu"),     # deeper representation of patterns
    layers.Dense(1, activation="sigmoid")    # output: probability of one class (binary classification)
])

# Compile the model
mlp_catdog.compile(
    optimizer="adam",                        # optimization algorithm
    loss="binary_crossentropy",              # loss for binary classification
    metrics=["accuracy"]                     # track accuracy during training
)

# summary of the MLP model architecture
mlp_catdog.summary()

```

```python
# train the MLP model
mlp_history = mlp_catdog.fit(
    train_ds_padded,
    validation_data=val_ds_padded,
    epochs=10
)

# evaluate the MLP model on test data
test_loss_mlp, test_acc_mlp = mlp_catdog.evaluate(test_ds_padded, verbose=2)
print(f"MLP Test accuracy: {test_acc_mlp:.4f}")
```

```python
# Plot training history for MLP (accuracy over epochs and loss over epochs)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(mlp_history.history['accuracy'], label='MLP Train Acc')
plt.plot(mlp_history.history['val_accuracy'], label='MLP Val Acc')
plt.title('MLP Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(mlp_history.history['loss'], label='MLP Train Loss')
plt.plot(mlp_history.history['val_loss'], label='MLP Val Loss')
plt.title('MLP Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

### Building CNN model

```python
# set random seed for reproducibility
tf.random.set_seed(42)

# define the CNN model
cnn_catdog = models.Sequential([
    layers.Input(shape=(128, 128, 3)),             # input: 128x128 RGB image

    layers.Conv2D(32, (3, 3), activation="relu",
                  padding="same"),                 # detect edges / simple patterns
    layers.MaxPooling2D((2, 2)),                   # reduce spatial size

    layers.Conv2D(64, (3, 3), activation="relu",
                  padding="same"),                 # detect more complex patterns
    layers.MaxPooling2D((2, 2)),                   # further reduce size

    layers.Flatten(),                              # convert feature maps to vector
    layers.Dense(64, activation="relu"),           # combine features
    layers.Dense(1, activation="sigmoid")          # binary output (cat vs dog)
])

# Compile CNN model
cnn_catdog.compile(
    optimizer="adam",                              # optimization algorithm
    loss="binary_crossentropy",                    # binary classification loss
    metrics=["accuracy"]                           # track accuracy
)

# Train CNN model
cnn_history = cnn_catdog.fit(
    train_ds_padded,                               # training dataset
    validation_data=val_ds_padded,                 # validation dataset
    epochs=20                                      # increase if needed
)

# Evaluate CNN on test set (unseen data)
cnn_test_loss, cnn_test_acc = cnn_catdog.evaluate(test_ds_padded)

print(f"CNN test accuracy: {cnn_test_acc:.4f}")
```

### CNN with early stopping

```python
# add early stopping to CNN model
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",                         # watch validation loss
    patience=5,                                 # stop if no improvement for 5 epochs
    restore_best_weights=True                   # revert to best model after stopping
)

# set random seed for reproducibility
tf.random.set_seed(42)

# define the CNN model
cnn_catdog = models.Sequential([
    layers.Input(shape=(128, 128, 3)),             # input: 128x128 RGB image

    layers.Conv2D(32, (3, 3), activation="relu",
                  padding="same"),                 # detect edges / simple patterns
    layers.MaxPooling2D((2, 2)),                   # reduce spatial size

    layers.Conv2D(64, (3, 3), activation="relu",
                  padding="same"),                 # detect more complex patterns
    layers.MaxPooling2D((2, 2)),                   # further reduce size

    layers.Flatten(),                              # convert feature maps to vector
    layers.Dense(64, activation="relu"),           # combine features
    layers.Dense(1, activation="sigmoid")          # binary output (cat vs dog)
])

# Compile CNN model
cnn_catdog.compile(
    optimizer="adam",                              # optimization algorithm
    loss="binary_crossentropy",                    # binary classification loss
    metrics=["accuracy"]                           # track accuracy
)

# Train CNN model
cnn_history = cnn_catdog.fit(
    train_ds_padded,                               # training dataset
    validation_data=val_ds_padded,                 # validation dataset
    epochs=20,                                     # increase if needed
    callbacks=[early_stopping]                     # add early stopping callback
)

# Evaluate CNN on test set (unseen data)
cnn_test_loss, cnn_test_acc = cnn_catdog.evaluate(test_ds_padded)

print(f"CNN test accuracy: {cnn_test_acc:.4f}")
```

```python
# visualizing train vs validation accuracy for CNN model
plt.figure(figsize=(12,5))
plt.subplot(1, 2, 1)
plt.plot(cnn_history.history["accuracy"], label="Train Accuracy")
plt.plot(cnn_history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CNN Accuracy (Train vs Validation)")
plt.legend()

# plot train vs validation loss for CNN model
plt.subplot(1, 2, 2)
plt.plot(cnn_history.history["loss"], label="Train Loss")
plt.plot(cnn_history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CNN Loss (Train vs Validation)")
plt.legend()

plt.show()

plt.show()
```

### Does CNN performance improve with data augmentation?

```python
# set random seed for reproducibility
tf.random.set_seed(42)

# define data augmentation
data_augmentation = models.Sequential([
    layers.RandomRotation(0.05),                # randomly rotate training images
    layers.RandomFlip("horizontal")             # randomly flip images left/right
])

# define CNN model with data augmentation
cnn_aug_catdog = models.Sequential([
    layers.Input(shape=(128, 128, 3)),          # input: 128x128 RGB image

    data_augmentation,                          # apply augmentation during training only

    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])

# compile augmented CNN model
cnn_aug_catdog.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# train augmented CNN model
cnn_aug_history = cnn_aug_catdog.fit(
    train_ds_padded,
    validation_data=val_ds_padded,
    epochs=20,
)

# evaluate augmented CNN on test set
cnn_aug_test_loss, cnn_aug_test_acc = cnn_aug_catdog.evaluate(test_ds_padded)

print(f"Augmented CNN test accuracy: {cnn_aug_test_acc:.4f}")
```

```python
# Model 1: CNN with Dropout + L2 Regularization
# Goal: reduce overfitting from baseline CNN using weight decay and dropout

tf.random.set_seed(42)

data_augmentation_m1 = models.Sequential([
    layers.RandomFlip("horizontal"),
])

early_stopping_m1 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

cnn_m1 = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    data_augmentation_m1,

    layers.Conv2D(32, (3, 3), padding="same",
                  kernel_regularizer=regularizers.l2(0.001)),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), padding="same",
                  kernel_regularizer=regularizers.l2(0.001)),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation="relu",
                 kernel_regularizer=regularizers.l2(0.001)),
    layers.Dropout(0.4),
    layers.Dense(1, activation="sigmoid")
])

cnn_m1.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

history_m1 = cnn_m1.fit(
    train_ds_padded,
    validation_data=val_ds_padded,
    epochs=20,
    callbacks=[early_stopping_m1]
)

loss_m1, acc_m1 = cnn_m1.evaluate(test_ds_padded)
print(f"Model 1 (L2 + Dropout) Test Accuracy: {acc_m1:.4f}")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history_m1.history["accuracy"], label="Train Acc")
plt.plot(history_m1.history["val_accuracy"], label="Val Acc")
plt.title("Model 1 Accuracy")
plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_m1.history["loss"], label="Train Loss")
plt.plot(history_m1.history["val_loss"], label="Val Loss")
plt.title("Model 1 Loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.legend()
plt.suptitle("Model 1: Dropout + L2 Regularization")
plt.tight_layout()
plt.show()
```

```python
# Model 2: Deeper CNN (3 Conv blocks) + Batch Normalization + Aggressive Augmentation
# Goal: increase model capacity and stabilize training with batch norm

tf.random.set_seed(42)

data_augmentation_m2 = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

early_stopping_m2 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

cnn_m2 = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    data_augmentation_m2,

    # Block 1
    layers.Conv2D(32, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(64, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 3 — added capacity
    layers.Conv2D(128, (3, 3), padding="same"),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(1, activation="sigmoid")
])

cnn_m2.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

history_m2 = cnn_m2.fit(
    train_ds_padded,
    validation_data=val_ds_padded,
    epochs=20,
    callbacks=[early_stopping_m2]
)

loss_m2, acc_m2 = cnn_m2.evaluate(test_ds_padded)
print(f"Model 2 (Deeper + BatchNorm + Augmentation) Test Accuracy: {acc_m2:.4f}")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history_m2.history["accuracy"], label="Train Acc")
plt.plot(history_m2.history["val_accuracy"], label="Val Acc")
plt.title("Model 2 Accuracy")
plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_m2.history["loss"], label="Train Loss")
plt.plot(history_m2.history["val_loss"], label="Val Loss")
plt.title("Model 2 Loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.legend()
plt.suptitle("Model 2: Deeper CNN + Batch Normalization + Strong Augmentation")
plt.tight_layout()
plt.show()
```

```python
# Model 3: Combined Kitchen-Sink Regularization (L2 + Dropout + BatchNorm + Full Augmentation)
# Goal: stack every regularization technique to find the performance ceiling on this small dataset

tf.random.set_seed(42)

data_augmentation_m3 = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomTranslation(0.05, 0.05),
    layers.RandomContrast(0.15),
])

early_stopping_m3 = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=6,
    restore_best_weights=True
)

cnn_m3 = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    data_augmentation_m3,

    # Block 1
    layers.Conv2D(64, (3, 3), padding="same",
                  kernel_regularizer=regularizers.l2(0.0005)),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    # Block 2
    layers.Conv2D(128, (3, 3), padding="same",
                  kernel_regularizer=regularizers.l2(0.0005)),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(128, activation="relu",
                 kernel_regularizer=regularizers.l2(0.0005)),
    layers.Dropout(0.5),
    layers.Dense(1, activation="sigmoid")
])

cnn_m3.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

history_m3 = cnn_m3.fit(
    train_ds_padded,
    validation_data=val_ds_padded,
    epochs=20,
    callbacks=[early_stopping_m3]
)

loss_m3, acc_m3 = cnn_m3.evaluate(test_ds_padded)
print(f"Model 3 (All Regularization) Test Accuracy: {acc_m3:.4f}")

# Summary comparison
print("\n--- Test Accuracy Summary ---")
print(f"Model 1 (L2 + Dropout):                      {acc_m1:.4f}")
print(f"Model 2 (Deeper + BatchNorm + Aug):           {acc_m2:.4f}")
print(f"Model 3 (L2 + Dropout + BatchNorm + Full Aug):{acc_m3:.4f}")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history_m3.history["accuracy"], label="Train Acc")
plt.plot(history_m3.history["val_accuracy"], label="Val Acc")
plt.title("Model 3 Accuracy")
plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_m3.history["loss"], label="Train Loss")
plt.plot(history_m3.history["val_loss"], label="Val Loss")
plt.title("Model 3 Loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.legend()
plt.suptitle("Model 3: Combined L2 + Dropout + BatchNorm + Full Augmentation")
plt.tight_layout()
plt.show()
```

| Demo aug CNN (flip + rotation, no ES) | ~83% | ~70% | **60.4%** |

### What actually happened (vs. what I expected)

The results were surprising. I expected stacking regularization techniques to progressively improve generalization, but the opposite occurred: **more regularization led to worse performance**, eventually pushing Model 3 below random chance (46.9%).

**Model 2** (3 conv blocks + BatchNorm + rotation/zoom/flip augmentation) barely improved beyond 50–55% training accuracy across 13 epochs — it essentially failed to learn. This is likely because combining aggressive augmentation with batch normalization on a dataset of only 800 images made the learning signal too noisy per batch. BatchNorm computes statistics per mini-batch, and with highly augmented, variable images, those statistics were unreliable.

**Model 3** (L2 + BatchNorm + Dropout + 5 augmentation types) achieved only 46.9% test accuracy — worse than a coin flip. Stacking L2 weight decay on top of an already-struggling model with BatchNorm and heavy augmentation caused extreme underfitting. The model was regularized so aggressively it could not fit even the training data.

### What actually helped

The **baseline CNN with no regularization** achieved the highest test accuracy (67.7%), despite training accuracy hitting 100% by epoch 11 — a textbook case of overfitting. The fact that it still generalized reasonably suggests the test set was not dramatically harder than the training set.

The **starter template** (horizontal flip only + early stopping) came in second at 65.6% while keeping the train/val gap much narrower. This represents the best actual generalization tradeoff: the model learned meaningful features without memorizing, and early stopping prevented the loss from diverging further.

### Key lesson: too much regularization is underfitting

On a dataset this small (800 training images), the primary bottleneck is **data quantity**, not model complexity. The models that failed (Models 2 and 3) were over-regularized — they were penalized so heavily they couldn't fit the training set, which guarantees poor test performance. The bias-variance tradeoff is a U-curve: moving too far in the regularization direction is just as harmful as moving too far toward overfitting.

The practical takeaway: **start simple, add one regularization technique at a time, and verify training accuracy stays well above 50%** before adding more constraints. Horizontal flip augmentation + early stopping was the highest-impact combination on this dataset. L2, BatchNorm, and heavy multi-augmentation all individually slowed or prevented learning when applied on top of each other.
